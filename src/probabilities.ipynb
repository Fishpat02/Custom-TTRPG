```python
import polars as pl

from ttrpg_testing import gen_dataframe

trials_frame = gen_dataframe()
```

```python
trials_agg = trials_frame.group_by(["level", "dc"], maintain_order=True).agg(
    pl.col("num_dice").max().name.replace("num", "max"),
    pl.mean("probability").name.suffix("_avg"),
    pl.median("probability").name.suffix("_mid"))

trials_agg
```

```python
import altair as alt

alt.theme.enable('dark')

trials_chart = (
    trials_agg
    .rename({"probability_avg": "Average", "probability_mid": "Median"})
    .plot.bar(x="level:O", color="dc:O")
    .transform_fold(["Median", "Average"])
    .properties(title="Probability Distribution")
    .encode(xOffset="dc:O", column=alt.Column("key:N", title=None),
            y=alt.Y("value:Q", title="Probability", scale=alt.Scale(domain=(0, 1))).axis(format="%")))

trials_chart
```
